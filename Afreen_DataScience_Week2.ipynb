```python
# TASK 1
# Arithmetic operations taking user inputs
a=int(input("Enter first number:"))    #assigning value1
b=int(input("Enter second number:"))   #assigning value2
add=a+b
sub=a-b
mul=a*b
div=a/b
exp=a**b
mod=a%b
print("\n *** The Arithmetic results are as follows***")
print(f"Addition",add)
print(f"Subtraction",sub)
print(f"Multiplication",mul)
print(f"Division",div)
print(f"Modulus",mod)
print(f"Exponential",exp)
```

```python
# TASK-2
# Working with Lists and Arrays

import numpy as np

#Creating a list containing 10 numbers
a=[14,24,37,48,51,62,78,89,93,22]
print("\n The list is",a)
print("\n The type of a is",type(a))
print("\n The length of the list is",len(a))
print("\n The maximum value in the list is",max(a))
print("\n The minimum value in the list is",min(a))

#Adding an element to the list
b=int(input("\n Enter the element to be added to the list:"))
a.append(b)
print("\n The updated list is",a)

# Removing an element from the list
c=int(input("\n Enter the element to be removed from the list:"))
a.remove(c)
print("\n The updated list after removing is\n",a)

# Printing in ascending and descending orders
print("\n The list in ascending order is",sorted(a))
print("\n The list in descending order is",sorted(a,reverse=True))

#Converting to numpy array
arr=np.array(a)
print("\n The type of a is now",type(arr))
print("\n The mean of the array is",np.mean(arr))
print("\n The median of the array is",np.median(arr))
print("\n The standard deviation of the array is",np.std(arr))


```

```python
# TASK-3
# Creating a dictionary named student with keys: name, age, course, marks
student={
    'name':'Afreen',
    'age':30,
    'course':'DataScience',
    'marks':95,
}
print("\n The dictionary is",student)

# Pairing key-value pairs using for-loop
for key,value in student.items():
    print(f"\n {key}:{value}")
print("\n Above are the key-value pairs")
# Adding a new key
student['grade']='A+'
print("\n The updated dictionary after adding grade key is",student)

# Creating a set of unique courses
courses={"Python", "Data Science", "AI", "Python"}
print("\n The set of unique courses is",courses)  #Set cant have duplicate values so Python wont get printed second time

#Set operations-union, intersection, and difference
set1={"Python", "Data Science", "AI"}
set2={"DataAnalyst","Python","AI","DataScientist", "AI Engineer"}
print("\n The set1 is",set1)
print("\n The set2 is",set2)
print("\n The union of set1 and set2 is",set1.union(set2))
print("\n The intersection of set1 and set2 is",set1.intersection(set2))
print("\n The difference between set1 and set2 is",set1.difference(set2))
```

```python
# TASK-4
# File Handling
# Creating a textfile with name,course and marks of 5 students
with open("student_data.txt","w") as file:
    file.write("Name,Course,Marks\n")
    file.write("Afreen,DataScience,95\n")
    file.write("John,DataAnalyst,90\n")
    file.write("Abel,DataScientist,85\n")
    file.write("Jane,AI,65\n")
    file.write("Mary,Python,73\n")

with open("student_data.txt","r") as file:
    data=file.read()   #displaying the file data
    print(data)

# Displaying the student name with marks>75
with open("student_data.txt","r") as file:
    lines=file.readlines()
    for line in lines[1:]:
        fields=line.strip().split(",")
        name=fields[0]
        marks=int(fields[2])
        if marks>75:
            print(name,"Scored more than 75\n")
```
