```python
# Importing required libraries
import nltk
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Downloading required datasets (run once)
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

# Input: Two paragraphs
text = """
Data science is an exciting field! It combines statistics, programming, and domain knowledge.
However, learning it requires patience, practice, and curiosity. Many people ask: "Is it difficult?"

Natural Language Processing (NLP) helps machines understand human language.
For example, chatbots, search engines, and translators use NLP techniques.
Amazing, right? Technology keeps evolving; we must keep learning!
"""

print("Original Text:\n")
print(text)
```

```python
# Removing punctuation
text_no_punct = text.translate(str.maketrans('', '', string.punctuation))

print("\nText after removing punctuation:\n")
print(text_no_punct)
```

```python
# Tokenization
import nltk
nltk.download('punkt_tab')
tokens = word_tokenize(text_no_punct)

print("\nTokenized Words:\n")
print(tokens)
```

```python
# Stop-word removal
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in tokens if word.lower() not in stop_words]

print("\nAfter Stopword Removal:\n")
print(filtered_words)
```

```python
# Stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("\nAfter Stemming:\n")
print(stemmed_words)
```

```python
# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("\nAfter Lemmatization:\n")
print(lemmatized_words)
```
